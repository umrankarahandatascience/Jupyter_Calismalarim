```python
# 5.Soru: 40 ile 100 sayısını hem ve hem de veya işlemlerine tabi tutarak sonucu yazdırınız.   
sayi1 = 40
sayi2 = 100

ve_sonuc = sayi1 & sayi2
veya_sonuc = sayi1 | sayi2

print("40 & 100 (VE Sonucu):", ve_sonuc)
print("40 | 100 (VEYA Sonucu:", veya_sonuc)
```

```python
# 10.Soru: X = 10 / 2 , Y = X + 5 , T = Y - 1 , P = T % 2  işlemleri tek satırda yazınız

P = (((10 // 2) + 5) - 1) % 2

print("P'nin değeri:", P)
```
