```python
# 12.Soru: Bir GSM operatörü müşterilerinin kaç yıllık abone olduğuna bakarak onlara hediye internet verecektir.
# Abone süresi 1 yıldan düşükse 1GB, 1-2 yıl arası 3GB, 2-3 yıl arası 5GB ve 3 yıldan fazla süredir abone olanlara 
# 8GB internet hediye edilmesi istenmektedir. Bu amaçla dışardan girilen abone süresine göre müşterinin kazanacağı 
# internet miktarını hesaplayınız.

yil = float(input("Kaç yıllık abone olduğunuzu giriniz: "))

if yil < 1:
    print("1GB hediye internet kazandınız.")

elif yil <= 2:
    print("3GB hediye internet kazandınız.")

elif yil <= 3:
    print("5GB hediye internet kazandınız.")
else:
    print("8GB hediye internet kazandınız.")
```

```python
# 13.Soru: if yapısını kullanarak x değişkeninin 30 sayısına göre durumunu (büyük,küçük veya eşit) bulunuz.

x = int(input(" x değerini giriniz: "))

if x > 30:
    print("Girdiğiniz x değeri 30 dan büyüktür.")

elif x < 30:
    print("Girdiğiniz x değeri 30 dan küçüktür.")
else:
    print("Girdiğiniz x değeri 30 a eşittir.")
```

```python
# 16.Soru: if yapısını kullanarak x=350 değişkeninin 100-200 arasında değilse,200-300 arasında olup olmadığını 
# birleştirme operatörlerini(ve,veya...) kullanmadan bulunuz.

x = 350

if x >= 100:
    if x <= 200:
        print(" x sayısı 100-200 arasındadır.")
    else:
        if x <= 300:
            print(" x sayısı 200-300 arasındadır.")
        else:
            print(" x sayısı 300'den büyüktür.")
else:
    print(" x sayısı 100'den küçüktür.")
```

```python
# 19.Soru: Dışarıdan girilen tutara ve cins verisine göre toplam tutarı belirleyiniz.
# Bir malın cinsi A ise %10,değilse %1 indirim uygulanacaktır.

tutar = float(input("Tutarı giriniz: "))
cins = input("Malın cinsini giriniz: ")

if cins == "A" or cins == "a":
    toplam_tutar = tutar - (tutar * 0.10)

else:
    toplam_tutar = tutar - (tutar * 0.01)

print("Ödenecek toplam tutar: ", toplam_tutar)
```

```python
# 20.Soru: Bir öğrencinin sözlü notunun olup olmadığını x mantıksal değişkeni ile belirlenmektedir. 
# a. X=ture ise sözlü olmuştur.  b. X=false ise sözlü olmamıştır.  
# Buna göre bir öğrencinin sözlü olup olmadığını ekrana tek satır şart ifadesi ile yazdırınız.

x = True

print("Sözlü olmuştur." if x else "Sözlü olmamıştır.")

x = False

print("Sözlü olmuştur." if x else "Sözlü olmamıştır.")
```

```python
# 24.Soru: Dışardan girilen 3 sayıdan en büyük 15 kaç sayı girildiğini bulunuz.

s1 = int(input("1.sayıyı giriniz: "))
s2 = int(input("2.sayıyı giriniz: "))
s3 = int(input("3.sayıyı giriniz: "))

sayac = 0

if s1 <= 15:
    sayac +=1
if s2 <= 15:
    sayac +=1

if s3 <= 15:
    sayac +=1

print("En büyük 15 olan ", sayac ," adet sayı girildi.")
```
